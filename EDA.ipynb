```python
# Create a crosstab of gender and addiction
crosstab = pd.crosstab(df['gender'], df['addicted_label'])
print(crosstab)

# Visualize the relationship
crosstab.plot(kind='bar', figsize=(10, 6))
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Relationship between Gender and Addiction')
plt.legend(title='Addicted Label')
plt.tight_layout()
plt.show()
```
